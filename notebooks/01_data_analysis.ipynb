```python
import os
import joblib

paths = [
    "data/processed/encoded.pkl",
    "../data/processed/encoded.pkl",
    "../../data/processed/encoded.pkl"
]

df = None

for path in paths:
    if os.path.exists(path):
        print(f"✅ Loading from: {path}")
        df = joblib.load(path)
        break

if df is None:
    raise FileNotFoundError("encoded.pkl not found. Run encoding pipeline first.")

# Memory safe
df = df.sample(n=min(10000, len(df)), random_state=42)

df.head()
```
